```python
import pandas as pd
import pathlib

# Path file input
file_path = pathlib.Path("dialogue_train.json")  # ganti dengan nama/path file Anda

# Direktori tempat file berada
base_dir = file_path

split_index = 400

# Membaca file JSON Lines
df = pd.read_json(file_path)

total_data = len(df)
print(f"Total data: {total_data}")

# Validasi split index
if split_index <= 0 or split_index >= total_data:
    raise ValueError(
        "split_index harus lebih besar dari 0 dan lebih kecil dari total data"
    )

# Split menjadi bagian kedua (data setelah index 400)
df_part2 = df.iloc[split_index:]

# Nama output file
output_2 = "dialogue_validation.json"

# Simpan hasil
df_part2.to_json(
    output_2,
    orient="records",
)

print(f"File berhasil disimpan: {output_2}")
```
