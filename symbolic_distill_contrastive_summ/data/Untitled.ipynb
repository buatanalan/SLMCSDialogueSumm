```python
import pandas as pd
import numpy as np

# 1. Membaca file JSON
file_path = 'all_dialogue_test_500_annotated.json'
df = pd.read_json(file_path, orient='records')

# 2. Ekstrak data dari kolom annotation
df['verdict'] = df['annotation'].apply(lambda x: x['verdict'])
df['corrected_summary'] = df['annotation'].apply(lambda x: x['corrected_summary'])

# 3. Update kolom summary berdasarkan kondisi
df['summary'] = np.where(df['verdict'] == 'revisi', df['corrected_summary'], df['summary'])

# 4. Hapus kolom bantuan yang tadi dibuat agar tidak mengotori file JSON asli
df = df.drop(columns=['verdict', 'corrected_summary'])

# 5. MENYIMPAN KEMBALI KE FILE JSON YANG SAMA
# Gunakan orient='records' dan force_ascii=False agar karakter khusus/non-ascii tidak rusak
df.to_json(file_path, orient='records', indent=4, force_ascii=False)

print(f"File berhasil diperbarui dan disimpan ke {file_path}")
```
