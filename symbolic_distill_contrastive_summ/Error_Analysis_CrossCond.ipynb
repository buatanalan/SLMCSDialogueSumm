# Error Analysis Lintas Kondisi — margin_part @ 250 vs 350 token

Karena **test set identik**, tiap dialog bisa dipasangkan antara kondisi max input 250 (method3)
dan 350 (method4). Dialog dikelompokkan berdasarkan perubahan AlignScore:

- **membaik** (Δalign ≥ +0.05) → error *truncation-induced*: hilang saat konteks diperpanjang.
- **persisten_rendah** (align tetap rendah di 350) → kelemahan model sebenarnya.
- **regresi** (Δalign ≤ −0.05) → align justru turun di 350 (kasus aneh, perlu dipahami).
- **stabil** → tidak banyak berubah.

Uji kunci: kalau truncation memang penyebabnya, kelompok **membaik harus lebih panjang** dari
kelompok lain. Sel ini menghitungnya dan menampilkan contoh tiap kelompok.
