```python
import pandas as pd

df = pd.read_json('data/dialogue_test.json', lines=True)
df.head()
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("Qwen/qwen2.5-0.5B")  # atau model kamu

df['dialogue_len_token'] = df['dialogue'].apply(
    lambda x: len(tokenizer.encode(x, add_special_tokens=False))
)

df['summary_len_token'] = df['summary'].apply(
    lambda x: len(tokenizer.encode(x, add_special_tokens=False))
)
```

```python
df.head()
```

```python
max_input_len = 256
max_output_len = 128

def truncate_text(text, max_len):
    tokens = tokenizer.encode(text, add_special_tokens=False)
    tokens = tokens[:max_len]
    return tokenizer.decode(tokens, skip_special_tokens=True)

df['dialogue'] = df['dialogue'].apply(lambda x: truncate_text(x, max_input_len))
df['summary'] = df['summary'].apply(lambda x: truncate_text(x, max_output_len))
```

```python
df[['dialogue','summary','summaries']].to_json(
    'data/dialogue_test.json',
    orient='records',
    lines=False,
    force_ascii=False
)
```
