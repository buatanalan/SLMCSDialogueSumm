# Cek Manual Terarah — Dialog, Reference, Prediksi & Skor

Sel ini menampilkan sampel dialog yang paling informatif untuk diperiksa manual, lengkap dengan
reference + prediksi + semua skor. **Tempelkan/jalankan setelah notebook analisis hasil Anda**
(membutuhkan variabel `res`, `metric_cols`, dan `texts` yang sudah ada di sana).

Empat kelompok sampel terarah:
1. **Dialog panjang + AlignScore terendah** — inti temuan: kenapa faktualitas turun saat dialog panjang.
2. **Domain taxi + AlignScore terendah** — memahami anomali taxi (lancar tapi kurang akurat).
3. **Fluency tinggi tapi AlignScore rendah** — kasus "terbaca bagus tapi salah".
4. **Skor rata-rata terendah & tertinggi** — sanity check apakah metrik masuk akal.

Jika kolom prediksi/reference tidak terdeteksi otomatis, set `PRED_COL`/`REF_COL` manual di atas sel.
Output juga diekspor ke `manual_check_samples.csv`.
